# BCI Motor Imagery Lab - Full Solution

**Subject S081 (PhysioNet BCI, runs R04/R08/R12), 64-ch, 160 Hz**

End-to-end pipeline: ICA -> 8-30 Hz bandpass -> CSP -> LDA/SVM -> Real-time streaming

## Task 1 - Preprocessing and ICA

- 4th-order Butterworth 8-30 Hz
- FastICA (15 components), ocular rejection by Fp1/Fp2 correlation
- 4-s epochs (tmin=0, tmax=4.0)


```python
# --- Task 1: load 9 EDFs, bandpass, ICA ocular rejection, epoching ---
import numpy as np
import mne
mne.set_log_level("WARNING")
import matplotlib

import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

BASE = r"C:\Users\PC\bci_practical"
RUNS = ["S083R04.edf", "S083R08.edf", "S083R12.edf"]
print(">>> starting")
montage = mne.channels.make_standard_montage("standard_1005")

raws = []
for fn in RUNS:
    r = mne.io.read_raw_edf(f"{BASE}\\{fn}", preload=True)
    r.rename_channels(lambda x: x.strip(". "))
    r.set_montage(montage, match_case=False, on_missing="ignore")
    r.annotations.description = np.array(
        [f"{d}_r{len(raws)}" for d in r.annotations.description]
    )
    print(f">>> {fn}: {r.info['nchan']} ch, {r.times[-1]:.1f} s")
    raws.append(r)

raw = mne.concatenate_raws(raws, verbose=False)
print(f">>> concatenated: {raw.times[-1]:.2f} s total")

# restore original T0/T1/T2 labels for the merged raw
raw.set_annotations(mne.Annotations(
    onset=raw.annotations.onset.copy(),
    duration=raw.annotations.duration.copy(),
    description=np.array(
        [d.split("_r")[0] for d in raw.annotations.description]
    ),
    orig_time=raw.annotations.orig_time,
))

# --- 1a. Bandpass 8-30 Hz (4th-order Butterworth) ---
raw_filt = raw.copy().filter(
    l_freq=8.0, h_freq=30.0,
    method="iir",
    iir_params=dict(order=4, ftype="butter", output="sos"),
    picks="eeg", verbose=False,
)
print(">>> bandpass 8-30 Hz done")

# --- 1b. ICA on 1-30 Hz copy ---
raw_ica = raw.copy().filter(l_freq=1.0, h_freq=30.0,
                            picks="eeg", verbose=False)
raw_ica.set_montage(montage, match_case=False, on_missing="ignore")
ica = mne.preprocessing.ICA(
    n_components=15, method="fastica",
    random_state=42, max_iter="auto",
)
ica.fit(raw_ica, picks="eeg")
ica.info.set_montage(montage, match_case=False, on_missing="ignore")
print(">>> ICA fitted:", ica.n_components_, "components")

fp_chs = [ch for ch in raw.ch_names if ch in ("Fp1", "Fp2")]
ica_src = ica.get_sources(raw_ica).get_data()
eog_src = raw_ica.copy().pick(fp_chs).get_data()
corrs = np.array([
    max(abs(np.corrcoef(ica_src[i], eog_src[j])[0, 1])
        for j in range(eog_src.shape[0]))
    for i in range(ica_src.shape[0])
])
eog_inds = np.where(corrs > 0.4)[0].tolist()
print(">>> ocular ICs to reject:", eog_inds,
      "| corr:", np.round(corrs[eog_inds], 2))
ica.exclude = eog_inds

raw_clean = raw_filt.copy()
ica.apply(raw_clean)
print(">>> ICA applied")

# SNR proxy: Fp variance reduction
fp_pre  = raw_filt.copy().pick(["Fp1", "Fp2"]).get_data()
fp_post = raw_clean.copy().pick(["Fp1", "Fp2"]).get_data()
snr_gain_db = 10 * np.log10(np.var(fp_pre) / np.var(fp_post))
print(f">>> Fp1/Fp2 variance reduction: {snr_gain_db:.2f} dB")

# --- 1c. Epoch 0-4 s ---
events, event_id = mne.events_from_annotations(
    raw_clean,
    event_id={"T0": 1, "T1": 2, "T2": 3},   # explicit mapping
    verbose=False,
)
print(">>> annotations:", event_id)
keep = {k: v for k, v in event_id.items() if k in ("T1", "T2")}
events_mi = events[np.isin(events[:, 2], list(keep.values()))]
event_id_mi = {"left": keep["T1"], "right": keep["T2"]}
print(f">>> events: {len(events_mi)} total, "
      f"{np.sum(events_mi[:,2]==keep['T1'])} left, "
      f"{np.sum(events_mi[:,2]==keep['T2'])} right")

epochs = mne.Epochs(
    raw_clean, events_mi, event_id_mi,
    tmin=0.0, tmax=4.0, baseline=None,
    preload=True, picks="eeg",
)
print(">>> epochs shape:", epochs.get_data().shape)

from collections import Counter
y = epochs.events[:, -1]
print(">>> class balance:", Counter(y.tolist()))

epochs.save(f"{BASE}\\epochs_clean-epo.fif", overwrite=True)
print(">>> saved -> epochs_clean-epo.fif")

# --- 1d. Report figures ---
try:
    fig_ica = ica.plot_components(picks=range(15), show=False)
    fig_ica.suptitle("ICA components (rejected ocular marked)", fontsize=10)
    fig_ica.savefig(f"{BASE}\\task1_ica_components.png",
                    dpi=120, bbox_inches="tight")
    plt.close(fig_ica)
    print(">>> saved task1_ica_components.png")
except Exception as e:
    print(">>> skipped ICA figure:", e)

sfreq = int(raw_filt.info["sfreq"])
t0 = int(events_mi[0, 0])
pre_data,  _ = raw_filt.copy().pick(["Fp1","Fp2"])[:, t0:t0+4*sfreq]
post_data, _ = raw_clean.copy().pick(["Fp1","Fp2"])[:, t0:t0+4*sfreq]

fig, axes = plt.subplots(2, 1, figsize=(10, 5), sharex=True)
for ax, data, title in zip(axes, [pre_data, post_data],
                           ["Before ICA", "After ICA"]):
    ax.plot(data[0], label="Fp1", lw=0.8)
    ax.plot(data[1], label="Fp2", lw=0.8)
    ax.set_title(title); ax.set_ylabel("uV")
    ax.legend(loc="upper right", fontsize=8)
axes[1].set_xlabel("samples @ 160 Hz")
fig.suptitle("Fp1/Fp2 segment - ICA effect on ocular artifact", fontsize=10)
plt.tight_layout()
plt.savefig(f"{BASE}\\task1_epochs_before_after.png", dpi=120)
plt.close(fig)
print(">>> saved task1_epochs_before_after.png")

print(">>> DONE")
```

## Task 2 - CSP, Topographies, PSD, Log-var scatter

- CSP (m=4), log-variance features
- Patterns + filters topographies
- Welch PSD (1-s Hamming, 50% overlap) at C3/C4
- Log-variance scatter of components 1 vs 2


```python
# --- Task 2: CSP, PSD, log-var scatter ---
import numpy as np
import mne
mne.set_log_level("ERROR")
import matplotlib

import matplotlib.pyplot as plt
from mne.decoding import CSP
from scipy.signal import welch
import warnings
warnings.filterwarnings("ignore")

BASE = r"C:\Users\PC\bci_practical"
print(">>> loading epochs")
epochs = mne.read_epochs(f"{BASE}\\epochs_clean-epo.fif", preload=True)
montage = mne.channels.make_standard_montage("standard_1005")
epochs.set_montage(montage, match_case=False, on_missing="ignore")

y = epochs.events[:, -1]
y_bin = (y == 3).astype(int)                     # 3 = right
print(">>> epochs:", epochs.get_data().shape,
      "| label counts:", np.bincount(y_bin, minlength=2))

X = epochs.get_data()

# --- 2a. CSP m=4 ---
csp = CSP(n_components=4, reg=None, log=True, norm_trace=False)
X_csp = csp.fit_transform(X, y_bin)
print(">>> CSP feature shape:", X_csp.shape)

var_per_comp = np.exp(X_csp).mean(axis=0)
lam_ratio = float(var_per_comp[0] / var_per_comp[-1])
print(">>> mean variance per CSP component:", np.round(var_per_comp, 3))
print(f">>> lambda ratio (m=4): {lam_ratio:.2f}")

# --- 2b. Topographies ---
try:
    fig_p = csp.plot_patterns(epochs.info, ch_type="eeg",
                              units="Patterns (AU)", size=1.5)
    fig_p.suptitle("CSP patterns: first 2 / last 2", fontsize=10)
    fig_p.savefig(f"{BASE}\\task2_csp_patterns.png",
                  dpi=120, bbox_inches="tight")
    plt.close(fig_p)
    print(">>> saved task2_csp_patterns.png")
except Exception as e:
    print(">>> skipped patterns:", e)

try:
    fig_f = csp.plot_filters(epochs.info, ch_type="eeg",
                             units="Filters (AU)", size=1.5)
    fig_f.suptitle("CSP filters", fontsize=10)
    fig_f.savefig(f"{BASE}\\task2_csp_filters.png",
                  dpi=120, bbox_inches="tight")
    plt.close(fig_f)
    print(">>> saved task2_csp_filters.png")
except Exception as e:
    print(">>> skipped filters:", e)

# --- 2c. PSD + ERD at C3 / C4 ---
sfreq = epochs.info["sfreq"]
freqs, psd = welch(X, fs=sfreq, window="hamming",
                   nperseg=int(sfreq*1.0), noverlap=int(sfreq*0.5),
                   axis=-1)
c3 = epochs.ch_names.index("C3")
c4 = epochs.ch_names.index("C4")
alpha = (freqs >= 8) & (freqs <= 12)

psd_left  = psd[y_bin == 0].mean(axis=0)
psd_right = psd[y_bin == 1].mean(axis=0)
aL_c3, aR_c3 = psd_left[c3][alpha].mean(), psd_right[c3][alpha].mean()
aL_c4, aR_c4 = psd_left[c4][alpha].mean(), psd_right[c4][alpha].mean()
erd_c3 = (aR_c3 - aL_c3) / (aR_c3 + aL_c3) * 100
erd_c4 = (aR_c4 - aL_c4) / (aR_c4 + aL_c4) * 100
print(f">>> ERD C3 (right vs left): {erd_c3:+.1f}%")
print(f">>> ERD C4 (right vs left): {erd_c4:+.1f}%")

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(freqs, 10*np.log10(psd_left[c3]),  label="C3 | left",  lw=1.5)
ax.plot(freqs, 10*np.log10(psd_right[c3]), label="C3 | right", lw=1.5)
ax.plot(freqs, 10*np.log10(psd_left[c4]),  label="C4 | left",  lw=1.5, ls="--")
ax.plot(freqs, 10*np.log10(psd_right[c4]), label="C4 | right", lw=1.5, ls="--")
ax.axvspan(8, 12, color="gold", alpha=0.15, label="alpha")
ax.axvspan(16, 24, color="skyblue", alpha=0.15, label="beta")
ax.set_xlabel("Frequency (Hz)"); ax.set_ylabel("Power (dB)")
ax.set_title("Welch PSD - SMR channels per class")
ax.legend(fontsize=9); ax.set_xlim(5, 30)
plt.tight_layout()
plt.savefig(f"{BASE}\\task2_psd_c3_c4.png", dpi=120)
plt.close(fig)
print(">>> saved task2_psd_c3_c4.png")

# --- 2d. Log-var scatter ---
fig, ax = plt.subplots(figsize=(5.5, 5))
ax.scatter(X_csp[y_bin==0,0], X_csp[y_bin==0,1],
           c="tab:blue", label="left", alpha=0.7, s=40)
ax.scatter(X_csp[y_bin==1,0], X_csp[y_bin==1,1],
           c="tab:red",  label="right", alpha=0.7, s=40)
ax.set_xlabel("CSP 1 (log-var)"); ax.set_ylabel("CSP 2 (log-var)")
ax.set_title("CSP log-variance separation"); ax.legend()
plt.tight_layout()
plt.savefig(f"{BASE}\\task2_csp_logvar_scatter.png", dpi=120)
plt.close(fig)
print(">>> saved task2_csp_logvar_scatter.png")
print(">>> DONE")
```

## Task 3 - LDA vs SVM + 10-fold CV

- LDA with Ledoit-Wolf shrinkage, SVM RBF with C in {0.1, 1, 10}, gamma in {scale, auto}
- 10-fold Stratified CV, ROC-AUC, confusion matrices


```python
# --- Task 3: classifier benchmark, 10-fold CV ---
import numpy as np
import mne
mne.set_log_level("ERROR")
import matplotlib

import matplotlib.pyplot as plt
from mne.decoding import CSP
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.svm import SVC
from sklearn.model_selection import (StratifiedKFold, cross_val_score,
                                     cross_val_predict, GridSearchCV)
from sklearn.metrics import (accuracy_score, roc_auc_score, roc_curve,
                             confusion_matrix, ConfusionMatrixDisplay)
import warnings
warnings.filterwarnings("ignore")

BASE = r"C:\Users\PC\bci_practical"
print(">>> loading epochs")
epochs = mne.read_epochs(f"{BASE}\\epochs_clean-epo.fif", preload=True)
y = epochs.events[:, -1]
y_bin = (y == 3).astype(int)
print(">>> sanity: labels should be ~50/50:", np.bincount(y_bin, minlength=2))
assert 15 < y_bin.sum() < 30, "class counts not 50/50 — check y_bin mapping"

X = epochs.get_data()
print(">>> X:", X.shape, "| label counts:", np.bincount(y_bin, minlength=2))
print(">>> first 10 y:", y[:10])
print(">>> first 10 y_bin:", y_bin[:10])

# Sanity check: correlation of labels with trial index (should be ~0)
idx_corr = np.corrcoef(np.arange(len(y_bin)), y_bin)[0, 1]
print(f">>> label/index correlation (should be ~0): {idx_corr:+.3f}")

pipe_lda = Pipeline([
    ("csp",    CSP(n_components=4, reg=None, log=True, norm_trace=False)),
    ("scaler", StandardScaler()),
    ("lda",    LinearDiscriminantAnalysis(solver="lsqr", shrinkage="auto")),
])
pipe_svm = Pipeline([
    ("csp",    CSP(n_components=4, reg=None, log=True, norm_trace=False)),
    ("scaler", StandardScaler()),
    ("svm",    SVC(kernel="rbf", probability=True, random_state=42)),
])

cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

acc_lda = cross_val_score(pipe_lda, X, y_bin, cv=cv,
                          scoring="accuracy", n_jobs=1)
print(f">>> LDA 10-fold: {acc_lda.mean()*100:.2f}% "
      f"± {acc_lda.std()*100:.2f}%")

param_grid = {"svm__C": [0.1, 1, 10],
              "svm__gamma": ["scale", "auto"]}
gs = GridSearchCV(pipe_svm, param_grid, cv=cv,
                  scoring="accuracy", n_jobs=1)
gs.fit(X, y_bin)
print(">>> SVM best params:", gs.best_params_)
print(f">>> SVM best CV acc: {gs.best_score_*100:.2f}%")

acc_svm = cross_val_score(gs.best_estimator_, X, y_bin, cv=cv,
                          scoring="accuracy", n_jobs=1)
print(f">>> SVM 10-fold (tuned): {acc_svm.mean()*100:.2f}% "
      f"± {acc_svm.std()*100:.2f}%")

y_prob_lda = cross_val_predict(pipe_lda, X, y_bin, cv=cv,
                               method="predict_proba", n_jobs=1)[:, 1]
y_prob_svm = cross_val_predict(gs.best_estimator_, X, y_bin, cv=cv,
                               method="predict_proba", n_jobs=1)[:, 1]
y_pred_lda = (y_prob_lda > 0.5).astype(int)
y_pred_svm = (y_prob_svm > 0.5).astype(int)

auc_lda = roc_auc_score(y_bin, y_prob_lda)
auc_svm = roc_auc_score(y_bin, y_prob_svm)
print(f">>> AUC LDA: {auc_lda:.3f}")
print(f">>> AUC SVM: {auc_svm:.3f}")

# Figure
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
fpr_l, tpr_l, _ = roc_curve(y_bin, y_prob_lda)
fpr_s, tpr_s, _ = roc_curve(y_bin, y_prob_svm)
axes[0].plot(fpr_l, tpr_l, label=f"LDA (AUC={auc_lda:.3f})", lw=2)
axes[0].plot(fpr_s, tpr_s, label=f"SVM (AUC={auc_svm:.3f})", lw=2)
axes[0].plot([0,1],[0,1],"k--", lw=1)
axes[0].set_xlabel("False Positive Rate")
axes[0].set_ylabel("True Positive Rate")
axes[0].set_title("ROC curves — 10-fold CV")
axes[0].legend(loc="lower right")

ConfusionMatrixDisplay(
    confusion_matrix(y_bin, y_pred_lda),
    display_labels=["left","right"],
).plot(ax=axes[1], colorbar=False)
axes[1].set_title(f"LDA (acc={accuracy_score(y_bin,y_pred_lda)*100:.1f}%)")

ConfusionMatrixDisplay(
    confusion_matrix(y_bin, y_pred_svm),
    display_labels=["left","right"],
).plot(ax=axes[2], colorbar=False)
axes[2].set_title(f"SVM (acc={accuracy_score(y_bin,y_pred_svm)*100:.1f}%)")

plt.tight_layout()
plt.savefig(f"{BASE}\\task3_benchmark.png", dpi=130)
plt.close(fig)
print(">>> saved task3_benchmark.png")

# Final fitted pipeline for task 4 reuse
pipe_lda.fit(X, y_bin)
print(">>> DONE")
```

## Task 4 - Real-time buffer + ITR

- 500 ms window, 62.5 ms step, 3-sample majority smoothing
- Per-chunk latency benchmark against 15 ms budget
- Wolpaw ITR


```python
# --- Task 4: real-time streaming buffer, latency, ITR ---
import numpy as np
import mne
mne.set_log_level("ERROR")
import time
import matplotlib

import matplotlib.pyplot as plt
from mne.decoding import CSP
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.metrics import accuracy_score
import warnings
warnings.filterwarnings("ignore")

BASE = r"C:\Users\PC\bci_practical"

# --- 1. Train 500-ms streaming model on subject 81 R04 ---
print(">>> loading raw and building 500-ms training windows")
montage = mne.channels.make_standard_montage("standard_1005")
raw = mne.io.read_raw_edf(f"{BASE}\\S081R04.edf", preload=True)
raw.rename_channels(lambda x: x.strip(". "))
raw.set_montage(montage, match_case=False, on_missing="ignore")
raw_filt = raw.copy().filter(8, 30, picks="eeg", verbose=False)

events, event_id = mne.events_from_annotations(
    raw_filt, event_id={"T0": 1, "T1": 2, "T2": 3}, verbose=False,
)
keep = {k: v for k, v in event_id.items() if k in ("T1","T2")}
events_mi = events[np.isin(events[:, 2], list(keep.values()))]
event_id_mi = {"left": keep["T1"], "right": keep["T2"]}

ep_rt = mne.Epochs(raw_filt, events_mi, event_id_mi,
                   tmin=0.0, tmax=0.5, baseline=None,
                   preload=True, picks="eeg")
X_rt = ep_rt.get_data()
y_rt = (ep_rt.events[:, -1] == event_id_mi["right"]).astype(int)
print(">>> 500-ms training set:", X_rt.shape,
      "| label counts:", np.bincount(y_rt, minlength=2))

pipe_rt = Pipeline([
    ("csp",    CSP(n_components=4, reg=None, log=True, norm_trace=False)),
    ("scaler", StandardScaler()),
    ("lda",    LinearDiscriminantAnalysis(solver="lsqr", shrinkage="auto")),
])
pipe_rt.fit(X_rt, y_rt)
acc_rt = accuracy_score(y_rt, pipe_rt.predict(X_rt))
print(f">>> training accuracy (500-ms windows): {acc_rt:.3f}")

# --- 2. Stream 500 ms / 62.5 ms step ---
sfreq = raw.info["sfreq"]
window_samples = int(0.500 * sfreq)      # 80
step_samples   = int(0.0625 * sfreq)     # 10

stream = raw_filt.get_data()
print(">>> stream:", stream.shape,
      "= ", stream.shape[1] / sfreq, "s")

latencies, predictions = [], []
history = []

for t in range(window_samples, stream.shape[1], step_samples):
    chunk = stream[:, t - window_samples : t][np.newaxis, ...]
    t0 = time.perf_counter()
    pred = int(pipe_rt.predict(chunk)[0])
    t1 = time.perf_counter()
    latencies.append((t1 - t0) * 1000.0)     # ms
    history.append(pred)
    smoothed = int(np.round(np.mean(history[-3:])))  # 3-sample majority
    predictions.append(smoothed)

latencies = np.array(latencies)
predictions = np.array(predictions)
print(f">>> chunks processed: {len(latencies)}")
print(f">>> latency mean {latencies.mean():.2f} ms | "
      f"median {np.median(latencies):.2f} | "
      f"P95 {np.percentile(latencies, 95):.2f} | "
      f"max {latencies.max():.2f}")
print(f">>> drops (>15 ms): {(latencies > 15).sum()}")

fig, ax = plt.subplots(figsize=(10, 3))
ax.plot(latencies, lw=0.8)
ax.axhline(15, color="red", ls="--", label="15 ms budget")
ax.set_xlabel("chunk index"); ax.set_ylabel("latency (ms)")
ax.set_title("Real-time pipeline latency per 62.5 ms step")
ax.legend(); plt.tight_layout()
plt.savefig(f"{BASE}\\task4_latency.png", dpi=120)
plt.close(fig)
print(">>> saved task4_latency.png")

# --- 3. Wolpaw ITR using CV accuracy ---
from sklearn.model_selection import cross_val_score, StratifiedKFold

n_min = int(np.bincount(y_rt, minlength=2).min())      # smaller class count
n_folds = min(5, n_min)                                 # safe fold count
print(f">>> using {n_folds}-fold CV on RT windows "
      f"(smallest class has {n_min} trials)")

cv_rt = StratifiedKFold(n_splits=n_folds, shuffle=True, random_state=42)
P = cross_val_score(pipe_rt, X_rt, y_rt, cv=cv_rt,
                    scoring="accuracy", n_jobs=1).mean()
print(f">>> CV accuracy on RT windows: {P:.3f}")

N = 2
if 0 < P < 1:
    B = (np.log2(N) + P*np.log2(P)
         + (1-P)*np.log2((1-P)/(N-1)))
else:
    B = 0.0
T = 0.5
ITR = B * (60 / T)
print(f">>> P = {P:.3f}  B = {B:.3f} bits/decision  "
      f"T = {T} s  ITR = {ITR:.2f} bits/min")
print(">>> DONE")
```
